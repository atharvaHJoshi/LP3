```python

import numpy as np
import matplotlib.pyplot as plt

# Gradient Descent function
def gradient_descent(derivative, x, lr, iterations):

    for i in range(iterations):
        x = x - lr * derivative(x)

    return x

# Original function
def f(x):
    return (x+3)**2

# Derivative
def derivative(x):
    return 2*(x+3)

# Find minimum
minimum = gradient_descent(derivative, 2, 0.1, 100)

print("Minimum x:", minimum)
print("Minimum y:", f(minimum))

# Plot graph
x = np.linspace(-6, 4, 100)
y = f(x)

plt.plot(x, y, label='y=(x+3)^2')
plt.scatter(minimum, f(minimum), color='red', label='Minimum')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid()
plt.show()
```
